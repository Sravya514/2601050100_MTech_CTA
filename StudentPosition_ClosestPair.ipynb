```python
import math


def distance(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)


def closest_pair(students):
    min_distance = float('inf')
    pair = None

    names = list(students.keys())
    points = list(students.values())

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            d = distance(points[i], points[j])

            if d < min_distance:
                min_distance = d
                pair = (names[i], names[j])

    return pair, min_distance


def main():

    students = {
        "Ravi": (2, 3),
        "Sita": (10, 15),
        "John": (4, 5),
        "Anu": (20, 25),
        "Rani": (8, 12)
    }

    pair, min_distance = closest_pair(students)

    print("--- Student Positions ---")

    for name, position in students.items():
        print(name, ":", position)

    print("\n--- Closest Students ---")
    print("Students:", pair[0], "and", pair[1])
    print("Distance:", round(min_distance, 2))


if __name__ == "__main__":
    main()
```
