```python
def best_weather_period(temperature):
    current = temperature[0]
    maximum = temperature[0]

    best_start = 0
    best_end = 0
    start = 0

    for i in range(1, len(temperature)):

        if current < 0:
            current = temperature[i]
            start = i
        else:
            current += temperature[i]

        if current > maximum:
            maximum = current
            best_start = start
            best_end = i

    return temperature[best_start:best_end + 1], maximum


def main():

    temperature = [3, -1, 4, 2, -2, 5, -6]

    period, total = best_weather_period(temperature)

    print("Temperature Changes:", temperature)
    print("Best Weather Period:", period)
    print("Maximum Change:", total)


if __name__ == "__main__":
    main()
```
